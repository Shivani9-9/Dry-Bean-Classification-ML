```python
# STEP 4: Data preparation and preprocessing

# STEP 4.1: IMPORT DATA PREPROCESSING LIBRARIES



import numpy as np

# Encode categorical target labels into numerical values
from sklearn.preprocessing import LabelEncoder

# Split dataset into training and testing subsets
from sklearn.model_selection import train_test_split

# Standardize numerical features
from sklearn.preprocessing import StandardScaler

print("\033[1m-Data preprocessing libraries imported successfully!\033[0m")



# STEP 4.2: DATA CLEANING AND MISSING VALUE HANDLING


# Create a working copy to preserve the original dataset
df_clean = df.copy()

# Calculate missing values in each column
missing_values = df_clean.isnull().sum()

print("\033[1m\n-Missing values before cleaning:\033[0m")
print(missing_values)

# Calculate the total number of missing values
total_missing = missing_values.sum()

# Handle missing values only when they exist
if total_missing > 0:

    # Identify numerical columns
    numerical_columns = df_clean.select_dtypes(
        include=np.number
    ).columns

    # Replace missing numerical values with column median
    for column in numerical_columns:
        df_clean[column] = df_clean[column].fillna(
            df_clean[column].median()
        )

    # Identify categorical columns
    categorical_columns = df_clean.select_dtypes(
        exclude=np.number
    ).columns

    # Replace missing categorical values with column mode
    for column in categorical_columns:

        if df_clean[column].isnull().any():
            df_clean[column] = df_clean[column].fillna(
                df_clean[column].mode()[0]
            )

    print("\033[1m-Missing values handled successfully!\033[0m")
    # Verify missing values after cleaning
    print("\033[1m-Missing values after cleaning:\033[0m")
    print(df_clean.isnull().sum())

    print("\033[1m-Total missing values after cleaning:\033[0m",
          df_clean.isnull().sum().sum())

else:
    print("\033[1m-No missing values found. Imputation is not required.\033[0m")





# STEP 4.3: DUPLICATE OBSERVATION HANDLING



# Count the number of duplicate observations
duplicate_count = df_clean.duplicated().sum()

print("\n\033[1m-Duplicate rows before cleaning:\033[0m", duplicate_count)

# Remove exact duplicate observations
if duplicate_count > 0:

    df_clean = df_clean.drop_duplicates().reset_index(drop=True)

    print("\033[1m-Duplicate observations removed successfully!\033[0m")
    # Display the updated dataset dimensions
    print("\033[1m-Dataset shape after cleaning:\033[0m", df_clean.shape)

    # Verify duplicate observations after cleaning
    print("\033[1m-Remaining duplicate rows:\033[0m", df_clean.duplicated().sum())

else:

    print("\033[1m-No duplicate observations found.\033[0m")





# STEP 4.4: FEATURE AND TARGET SEPARATION



# Extract the input features by removing the target column
X = df_clean.drop(columns=["Class"])

# Extract the target variable containing bean classifications
y = df_clean["Class"]

# Display the dimensions of input features and target
print("\n\033[1m-Input feature dimensions:\033[0m", X.shape)
print("\033[1m-Target variable dimensions:\033[0m", y.shape)

# Display all input feature names
print("\033[1m-Input feature names:\033[0m")
print(X.columns.tolist())

# Display unique target classes
print("\033[1m-Target classes:\033[0m")
print(y.unique())

# Display the number of unique target classes
print("\033[1m-Number of target classes:\033[0m", y.nunique())




# STEP 4.5: TARGET LABEL ENCODING



# Initialize the LabelEncoder
label_encoder = LabelEncoder()

# Convert categorical target labels into numerical values
y_encoded = label_encoder.fit_transform(y)

# Display the original class-to-numerical mapping
print("\n\033[1m-Target label encoding:\033[0m")

for numerical_value, class_name in enumerate(
    label_encoder.classes_
):
    print(class_name, "=", numerical_value)

# Display the first ten encoded target values
print("\033[1m-First ten encoded target values:\033[0m")
print(y_encoded[:10])

# Display the total number of encoded classes
print("\033[1m-Total encoded classes:\033[0m", len(label_encoder.classes_))



# STEP 4.6: FEATURE ENGINEERING



# Create a separate copy of the input feature dataset
X = X.copy()

# Calculate aspect ratio using major and minor axis lengths
X["Aspect_Ratio"] = (
    X["MajorAxisLength"] / X["MinorAxisLength"]
)

# Display the original measurements and engineered feature
print("\n\033[1mEngineered feature: Aspect_Ratio\033[0m")

display(
    X[[
        "MajorAxisLength",
        "MinorAxisLength",
        "Aspect_Ratio"
    ]].head()
)

# Display updated feature dimensions
print("\033[1m-Updated feature dimensions:\033[0m", X.shape)

print("\033[1m-Total input features after feature engineering:\033[0m",
      X.shape[1])




# STEP 4.7: TRAIN-TEST DATA SPLITTING



# Split the dataset into training and testing subsets
X_train, X_test, y_train, y_test = train_test_split(

    X,                      # Input features
    y_encoded,              # Encoded target labels

    test_size=0.20,         # 20% testing and 80% training

    random_state=42,        # Reproducible random splitting

    stratify=y_encoded      # Preserve class proportions
)

# Display training and testing dataset dimensions
print("\n\033[1m-Training feature dimensions:\033[0m", X_train.shape)
print("\033[1m-Testing feature dimensions:\033[0m", X_test.shape)

print("\033[1m-Training target dimensions:\033[0m", y_train.shape)
print("\033[1m-Testing target dimensions:\033[0m", y_test.shape)

# Calculate training and testing percentages
print("\033[1m-Training percentage:\033[0m",
      round(len(X_train) / len(X) * 100, 2), "%")

print("\033[1m-Testing percentage:\033[0m",
      round(len(X_test) / len(X) * 100, 2), "%")




# STEP 4.8: FEATURE SCALING USING STANDARD SCALER



# Initialize the StandardScaler
scaler = StandardScaler()

# Learn scaling parameters and transform training features
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling parameters to testing features
X_test_scaled = scaler.transform(X_test)

# Convert scaled arrays into DataFrames
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)

# Display scaled training observations
print("\033[1m\n-Scaled training dataset:\033[0m")
display(X_train_scaled.head())

# Verify the mean and standard deviation of scaled features
print("\033[1m-Mean of scaled training features:\033[0m")
print(X_train_scaled.mean().round(3))

print("\033[1m-Standard deviation of scaled training features:\033[0m")
print(X_train_scaled.std().round(3))

print("\033[1m-Feature scaling completed successfully!\033[0m")
```

```python


# STEP 5: FINAL DATA PREPARATION VERIFICATION



print("\033[1m=" * 60)
print("       FINAL DATA PREPARATION SUMMARY")
print("=" * 60)

print("\n1. DATASET INFORMATION\033[0m")

print("Total observations:", len(df_clean))
print("Original features:", 16)
print("Engineered features:", X.shape[1])
print("Target classes:", len(label_encoder.classes_))

print("\033[1m\n2. DATA SPLITTING\033[0m")

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("\033[1m\n3. FEATURE SCALING\033[0m")

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled testing shape:", X_test_scaled.shape)

print("\033[1m\n4. DATA QUALITY\033[0m")

print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())

print("\033[1m\n5. TARGET CLASSES\033[0m")

print(label_encoder.classes_)

print("\033[1m\n" + "=" * 60)
print("DATA PREPARATION COMPLETED SUCCESSFULLY!")
print("=" * 60)
print("\033[0m")
```

```python
# STEP 6: Machine Learning Model Development

# STEP 6.1: IMPORT MACHINE LEARNING MODELS AND METRICS



# Import Logistic Regression classifier
from sklearn.linear_model import LogisticRegression

# Import Decision Tree classifier
from sklearn.tree import DecisionTreeClassifier

# Import K-Nearest Neighbors classifier
from sklearn.neighbors import KNeighborsClassifier

# Import Gaussian Naive Bayes classifier
from sklearn.naive_bayes import GaussianNB

# Import Support Vector Machine classifier
from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV

# Import evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

# Import time to measure model training duration
import time

print("\033[1mAll five ML models and evaluation metrics imported successfully!\033[0m")



# STEP 6.2: INITIALIZE FIVE CLASSIFICATION MODELS



# 1. Logistic Regression
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# 2. Decision Tree Classifier
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# 3. K-Nearest Neighbors Classifier
knn_model = KNeighborsClassifier(
    n_neighbors=5
)

# 4. Gaussian Naive Bayes Classifier
naive_bayes_model = GaussianNB()

# 5. Support Vector Machine Classifier
svm_model = CalibratedClassifierCV(
    estimator=SVC(kernel="rbf", random_state=42),
    method="sigmoid",
    cv=5
)

# Store all five models in a dictionary
models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "K-Nearest Neighbors": knn_model,
    "Gaussian Naive Bayes": naive_bayes_model,
    "Support Vector Machine": svm_model
}

# Display initialized models
print("\n\033[1mFive classification models initialized successfully!\033[0m")

for model_name, model in models.items():
    print("-",model_name, ":", model)




# STEP 6.3: PREPARE DATA FOR EACH CLASSIFICATION MODEL


# Store the appropriate training and testing data for each model

model_data = {

    "Logistic Regression": (
        X_train_scaled,
        X_test_scaled
    ),

    "Decision Tree": (
        X_train,
        X_test
    ),

    "K-Nearest Neighbors": (
        X_train_scaled,
        X_test_scaled
    ),

    "Gaussian Naive Bayes": (
        X_train,
        X_test
    ),

    "Support Vector Machine": (
        X_train_scaled,
        X_test_scaled
    )
}

# Display the dataset configuration for each model
print("\033[1m\nTraining and testing datasets assigned successfully!\033[0m")

for model_name, (train_data, test_data) in model_data.items():

    print("\033[1m",model_name,"\033[0m")
    print("Training shape:", train_data.shape)
    print("Testing shape:", test_data.shape)
    print("-" * 45,"\n")



# STEP 6.4: TRAIN ALL FIVE CLASSIFICATION MODELS


# Dictionary to store training execution times
training_times = {}

# Train each classification model
for model_name, model in models.items():

    # Retrieve the appropriate training dataset
    X_train_model, X_test_model = model_data[model_name]

    # Record the starting time
    start_time = time.time()

    # Train the model using training data
    model.fit(X_train_model, y_train)

    # Calculate training execution time
    end_time = time.time()

    training_time = end_time - start_time

    # Store the execution time
    training_times[model_name] = training_time

    # Display training status
    print("\033[1mModel:", model_name,"\033[0m")
    print("Training completed successfully!")
    print("Training time:", round(training_time, 4), "seconds")

print("\033[1mAll five classification models trained successfully!\033[0m")
```

```python
# STEP 7: MODEL EVALUATION AND CMPARISION


# STEP 7.1: MODEL PREDICTION AND PERFORMANCE EVALUATION


# Dictionaries to store predictions, probabilities and evaluation results
predictions = {}
probabilities = {}
evaluation_results = {}

# Evaluate each trained classification model
for model_name, model in models.items():

    # Retrieve the appropriate testing dataset
    X_test_model = model_data[model_name][1]

    # Generate predicted class labels
    y_pred = model.predict(X_test_model)

    # Generate class probabilities for AUC calculation
    y_prob = model.predict_proba(X_test_model)

    # Store predictions and probabilities
    predictions[model_name] = y_pred
    probabilities[model_name] = y_prob

    # Calculate Accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Calculate multiclass AUC using One-vs-Rest
    auc = roc_auc_score(
        y_test,
        y_prob,
        multi_class="ovr",
        average="weighted"
    )

    # Calculate weighted Precision
    precision = precision_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    # Calculate weighted Recall
    recall = recall_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    # Calculate weighted F1-score
    f1 = f1_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    # Calculate Matthews Correlation Coefficient
    mcc = matthews_corrcoef(y_test, y_pred)

    # Store all six evaluation metrics
    evaluation_results[model_name] = {
        "Accuracy": accuracy,
        "AUC": auc,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "MCC": mcc
    }

    # Display evaluation results for the current model
    print("\033[1m\n" + "=" * 55)
    print("MODEL:", model_name)
    print("=" * 55,"\033[0m")

    print("Accuracy :", round(accuracy, 4))
    print("AUC      :", round(auc, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1-Score :", round(f1, 4))
    print("MCC      :", round(mcc, 4))

print("\033[1mAll five models evaluated successfully!\033[0m")




# STEP 7.2: CREATE MODEL COMPARISON TABLE


# Convert evaluation results into a Pandas DataFrame
results_df = pd.DataFrame(evaluation_results).T

# Add training execution time
results_df["Training Time (seconds)"] = pd.Series(training_times)

# Round metric values for better readability
results_df = results_df.round(4)

# Display the final comparison table
print("\n\033[1mMODEL PERFORMANCE COMPARISON\033[0m")

display(results_df)




# STEP 7.3: IDENTIFY THE BEST-PERFORMING MODEL

# Find the model with the highest accuracy
best_model_name = results_df["Accuracy"].idxmax()

# Retrieve the best model's evaluation results
best_model_results = results_df.loc[best_model_name]

print("\033[1m\nBEST-PERFORMING MODEL\033[0m")
print("-" * 40)

print("\033[1mModel Name:", best_model_name,"\033[0m")

print("Performance Metrics:")
display(best_model_results)





# STEP 7.4: VISUAL COMPARISON OF ALL FIVE ML MODELS

import seaborn as sns

# Professional colour palette
model_colors = [
    "#264653",
    "#2A9D8F",
    "#E9C46A",
    "#F4A261",
    "#E76F51"
]

metric_colors = [
    "#264653",
    "#2A9D8F",
    "#E9C46A",
    "#F4A261",
    "#E76F51",
    "#7B61A8"
]


# GRAPH 1: MODEL ACCURACY COMPARISON


ranked_models = results_df.sort_values(
    by="Accuracy",
    ascending=False
)

plt.figure(figsize=(11, 6))

bars = plt.bar(
    ranked_models.index,
    ranked_models["Accuracy"],
    color=model_colors,
    edgecolor="white",
    linewidth=1.5,
    width=0.65
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.015,
        f"{height:.3f}",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

plt.title(
    "\nAccuracy Comparison of Five ML Models",
    fontsize=15,
    fontweight="bold",
    pad=18
)

plt.xlabel("Machine Learning Models", fontsize=12)
plt.ylabel("Accuracy Score", fontsize=12)

plt.xticks(rotation=20, ha="right")
plt.ylim(0, 1.1)

plt.grid(axis="y", linestyle="--", alpha=0.25)
plt.gca().set_axisbelow(True)

plt.tight_layout()
plt.show()


# GRAPH 2: PERFORMANCE COMPARISON HEATMAP


metrics = [
    "Accuracy",
    "AUC",
    "Precision",
    "Recall",
    "F1-Score",
    "MCC"
]

metric_data = results_df[metrics]

plt.figure(figsize=(12, 6))

sns.heatmap(
    metric_data,
    annot=True,
    fmt=".3f",
    cmap="BuPu",
    linewidths=1,
    linecolor="white",
    vmin=-1,
    vmax=1,
    cbar_kws={"label": "Performance Score"}
)

plt.title(
    "\nPerformance Comparison of Five ML Models",
    fontsize=15,
    fontweight="bold",
    pad=18
)

plt.xlabel("Evaluation Metrics", fontsize=12)
plt.ylabel("Machine Learning Models", fontsize=12)

plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()


# GRAPH 3: BEST MODEL FOR EACH PERFORMANCE METRIC


print("\n\033[1mBEST MODEL FOR EACH PERFORMANCE METRI\033[0m")

for metric in metrics:

    best_name = results_df[metric].idxmax()
    best_score = results_df[metric].max()

    print(f"{metric}: {best_name} ({best_score:.4f})")


# GRAPH 4: OVERALL PERFORMANCE COMPARISON


ax = metric_data.plot(
    kind="bar",
    figsize=(14, 7),
    width=0.8,
    color=metric_colors,
    edgecolor="white",
    linewidth=0.8
)

plt.title(
    "\nOverall Performance Comparison of ML Models",
    fontsize=16,
    fontweight="bold",
    pad=18
)

plt.xlabel("Machine Learning Models", fontsize=12)
plt.ylabel("Performance Score", fontsize=12)

# Include negative and positive values for MCC
plt.ylim(-1.0, 1.1)

# Highlight the zero reference line
plt.axhline(
    y=0,
    color="#333333",
    linewidth=1.3
)

plt.yticks(np.arange(-1.0, 1.1, 0.2))

plt.xticks(rotation=20, ha="right")

plt.legend(
    title="Evaluation Metrics",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=False
)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

```python
# STEP 8: CONFUSION MATRIX AND CLASSIFICATION REPORT


# STEP 8.1: GENERATE CONFUSION MATRICES FOR ALL FIVE MODELS


from sklearn.metrics import ConfusionMatrixDisplay

# Get the encoded class labels and their original names
class_labels = np.arange(len(label_encoder.classes_))
class_names = label_encoder.classes_

# Create a figure with space for five confusion matrices
fig, axes = plt.subplots(2, 3, figsize=(20, 12))

# Flatten the axes to make iteration easier
axes = axes.flatten()

# Display a confusion matrix for each trained model
for i, (model_name, y_pred) in enumerate(predictions.items()):

    # Calculate confusion matrix
    cm = confusion_matrix(
        y_test,
        y_pred,
        labels=class_labels
    )

    # Display the confusion matrix
    display_obj = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=class_names
    )

    display_obj.plot(
        ax=axes[i],
        cmap="YlGnBu",
        colorbar=False,
        values_format="d",
        xticks_rotation=45
    )

    # Add model name as the subplot title
    axes[i].set_title(
        model_name,
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    axes[i].set_xlabel("Predicted Class")
    axes[i].set_ylabel("Actual Class")

# Remove the unused sixth subplot
fig.delaxes(axes[5])

# Add an overall heading
fig.suptitle(
    "Confusion Matrix Comparison of Five ML Models",
    fontsize=18,
    fontweight="bold",
    y=1.02
)

plt.tight_layout()
plt.show()




# STEP 8.2: GENERATE CLASSIFICATION REPORTS



# Generate a classification report for each model
for model_name, y_pred in predictions.items():

    print("\n\033[1m")
    print("=" * 65)

    print("CLASSIFICATION REPORT:", model_name)

    print("=" * 65)
    print("\033[0m")

    # Generate and display the classification report
    report = classification_report(
        y_test,
        y_pred,
        labels=np.arange(len(label_encoder.classes_)),
        target_names=label_encoder.classes_,
        zero_division=0
    )

    print(report)
```

```python
# Lasso Regression using Gradient Descent
def lasso_regression(X, y, alpha, learning_rate=0.001, iterations=5000):

    weights = np.zeros(X.shape[1])

    for _ in range(iterations):

        prediction = X @ weights
        error = prediction - y

        gradient = (X.T @ error) / len(y)

        # Apply L1 penalty except to intercept
        gradient[1:] += alpha * np.sign(weights[1:])

        weights -= learning_rate * gradient

    return weights
```

```python
lasso_alphas = [0.0001, 0.001, 0.01, 0.1, 1]

lasso_original_results = []

for alpha in lasso_alphas:

    weights = lasso_regression(
        np.c_[np.ones(len(X_train)), X_train],
        y_train,
        alpha
    )

    log_pred = np.c_[
        np.ones(len(X_val)), X_val
    ] @ weights

    y_pred = np.maximum(
        np.expm1(log_pred), 0
    )

    score = rmsle(y_actual, y_pred)

    lasso_original_results.append([alpha, score])


# Create the result table
lasso_original_results = pd.DataFrame(
    lasso_original_results,
    columns=['Alpha', 'RMSLE']
)

# Add Sr.No.
lasso_original_results.insert(
    0,
    'Sr.No.',
    range(1, len(lasso_original_results) + 1)
)

# Round RMSLE
lasso_original_results['RMSLE'] = lasso_original_results['RMSLE'].round(4)

# Display without pandas index
display(
    lasso_original_results.style.hide(axis='index')
)
```

### Q10. Why does RMSLE penalize under-predictions more gently than RMSE?

RMSLE uses a **logarithm** when calculating the error, while RMSE uses the actual difference between predicted and actual values.

Because of the log transformation, very large errors have a smaller effect on RMSLE. It mainly looks at the **relative difference** between the actual and predicted values.

For example, predicting 50 when the actual value is 100 is treated less harshly by RMSLE than by RMSE.

Therefore, RMSLE is useful for this bike-sharing problem because the number of rentals can vary from small values to very large values.

### Q11. What are the trade-offs between model simplicity and predictive power?

A **simple model** is easy to understand, fast to train, and less likely to overfit the data. However, it may not be able to capture complex patterns, so its predictions may be less accurate.

A **more complex model** can capture more patterns in the data and may give better predictions. But it is harder to understand, takes more time to train, and may **overfit** the training data.

Therefore, the goal is to find a good balance between **simplicity and predictive performance**. In this assignment, we compare different models using RMSLE and select the model that gives the **lowest validation RMSLE**.

### Q12. Why can’t Linear Regression alone capture time-of-day effects effectively?

Linear Regression assumes that the relationship between a feature and the target is mostly **straight and linear**.

However, bike demand changes differently at different hours. For example, demand may be low at night, increase during the morning, decrease during the afternoon, and increase again in the evening.

So, treating `hour` as a simple number from 0 to 23 cannot represent these different patterns properly.

Therefore, Linear Regression alone cannot capture the **complex and non-linear effect of time of day** effectively. Extra features such as hour categories, cyclical features, or polynomial features can help the model capture these patterns better.
